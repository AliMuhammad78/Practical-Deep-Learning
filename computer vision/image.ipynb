# OpenCV and Images 

```python
import numpy as np 
import matplotlib.pyplot as plt
import cv2 as cv 
```

```python
img = cv.imread('./images/abc1234.jpg')
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# BGR (OpenCV default) ko RGB (Matplotlib default) mein badlein
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Ab is converted image ko dikhayein
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

### Convert the image to gray scale 

```python
gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
plt.imshow(gray_img, cmap='gray')
plt.axis('off')
plt.show()
```

## Read webcam using csv 

### Use in separate py file for this not notebook

#### But wsl can not access open your camera 

```python
import cv2 as cv
import matplotlib.pyplot as plt
from IPython.display import clear_output

# 1. Webcam access karein (WSL mein hamesha index check karna padta hai, 0 ya 1)
cap = cv.VideoCapture(0) 

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("Webcam se frame nahi mil raha!")
            break

        # 2. BGR se RGB convert karein (Matplotlib ke liye)
        frame_rgb = cv.cvtColor(frame, cv.COLOR_BGR2RGB)

        # 3. Purani image clear karein aur nayi display karein
        clear_output(wait=True)
        plt.imshow(frame_rgb)
        plt.axis('off')
        plt.show()

        # Loop ko manually stop karne ke liye  
        # Ya phir ye trick:
except KeyboardInterrupt:
    print("Video Stream Stopped.")

finally:
    cap.release()
    print("Camera Released.")
```
